# Tkinter Text Editor Bug Fix


```python
import tkinter as tk
from tkinter import ttk, font, colorchooser, messagebox, filedialog
import os
```

```python
def color_change():
    color = colorchooser.askcolor()[1]
    if color:
        text_editor.config(fg=color)
```

# Error 

config(fg=...) changes the foreground color of the entire Text widget instead of the selected text.
The use of text_editor.config(fg=color) indicated that the foreground color was being set on the entire Text widget. Since config() changes widget properties, it affects all text contained within the editor instead of only the selected portion.

# Solution

```python
def color_change():
    color = colorchooser.askcolor()[1]

    if color:
        tag_name = f"color_{color}"
        text_editor.tag_configure(tag_name, foreground=color)
        toggle_tag(tag_name)
```

# How I decided on the solution

Before writing a new solution, I examined the rest of the program to understand how other formatting features were implemented. I noticed that Bold, Italic, and Underline already worked correctly on selected text.

For example:

def change_bold():
    toggle_tag("bold")

which uses:

def toggle_tag(tag):
    ...
    text_editor.tag_add(tag, start, end)

This showed that the application was already using Tkinter Text Tags for selective formatting. Instead of creating a completely new implementation, I decided to follow the same design pattern.

# Why I used toggle_tag()

I reused the existing toggle_tag() function because it was already designed to apply formatting only to the selected text. Reusing existing functionality keeps the code consistent, avoids duplication, and follows good software engineering practices.

# Why I configured a color tag first

Unlike bold, italic, and underline, colors are not fixed formatting options. Every color selected from the color picker is different. Therefore, I first created a tag for the selected color using tag_configure(), and then passed that tag to toggle_tag() so it could be applied to the selected text.

# Result

After implementing this change, only the selected text changes color while the remaining document remains unchanged. This behavior matches the expected functionality of modern text editors such as Microsoft Word and Google Docs
